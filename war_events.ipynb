```python
import re

date_pattern = re.compile(
    r"\b(\d{1,2}\s+(January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{4})\b"
)

range_header_pattern = re.compile(
    r"^\d+\s+(?=(\d{1,2}\s+\w+|\w+)\s+[-–])"
)

def clean_text_list(lines):
    cleaned = []

    for line in lines:
        line = line.strip()
        if not line:
            continue

        # Remove page numbers: lines that ONLY contain digits
        if re.fullmatch(r"\d{1,3}", line):
            continue

        # Remove numeric prefix in date-range headers
        # e.g., "1 24 February ..." → "24 February ..."
        line = range_header_pattern.sub("", line)

        # Normalize internal spaces
        line = re.sub(r"\s+", " ", line).strip()

        cleaned.append(line)

    return cleaned
```

```python
cleaned_list = clean_text_list(text_list)
```

```python
cleaned_list = [s for s in cleaned_list if s not in summary_list]
cleaned_list[:100]
```

```python
def parse_key_events(text_list):
    date_pattern = re.compile(r'^\d{1,2}\s+\w+\s+\d{4}:')

    timeline = {}
    current_date = None
    current_text = []

    for line in text_list:
        line = line.strip()

        if not line or line == 'Key events':
            continue
        if date_pattern.match(line):
            if current_date:
                timeline[current_date] = ' '.join(current_text).strip()

            parts = line.split(':', 1)
            current_date = parts[0].strip()
            current_text = [parts[1].strip()] if len(parts) > 1 else []
        else:
            if current_date:
                current_text.append(line)

    if current_date:
        timeline[current_date] = ' '.join(current_text).strip()

    return timeline

key_events = parse_key_events(cleaned_list)
print('*'*8, 'Dictionary of Key Events', '*'*8)
for date, description in key_events.items():
    print(f"{date}:")
    print(f"  {description}")
    print()
```

```python
def parse_timeline_detailed(text_list):
    """Parse DETAILED timeline (format: 'date Description...')"""

    date_pattern = re.compile(r'^\d{1,2}\s+\w+\s+\d{4}\s+[A-Z]')

    timeline = {}
    current_date = None
    current_text = []

    for line in text_list:
        line = line.strip()

        if not line:
            continue

        if date_pattern.match(line):
            if current_date:
                timeline[current_date] = ' '.join(current_text).strip()

            parts = line.split(None, 3)
            if len(parts) >= 4:
                current_date = ' '.join(parts[:3])
                current_text = [parts[3]]
            else:
                current_date = ' '.join(parts)
                current_text = []
        else:
            if current_date:
                current_text.append(line)
    if current_date:
        timeline[current_date] = ' '.join(current_text).strip()

    return timeline
```

```python
detailed_dict = parse_timeline_detailed(cleaned_list)
print('*'*8, 'Dictionary of Detailed Events', '*'*8)
for date, description in detailed_dict.items():
    print(f"{date}:")
    print(f"  {description[:100]}...")  # First 100 chars
    print('\nlength: ', len(description))
    print()
```

```python
import csv

fieldnames = ['date', 'description']
with open('key_events.csv', 'w', newline='') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader() # Write the header row

    # Write the key-value pairs as rows
    for date, description in key_events.items():
        writer.writerow({'date': date, 'description': description})

with open('all_events.csv', 'w', newline='') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader() # Write the header row

    # Write the key-value pairs as rows
    for date, description in detailed_dict.items():
        writer.writerow({'date': date, 'description': description})
```
